```python
import pandas as pd
import os

DATA_DIR = "/kaggle/input/competitions/rsna-knee-abnormality-detection"

train = pd.read_csv(DATA_DIR + "/train.csv")
train_series = pd.read_csv(DATA_DIR + "/train_series.csv")

# First 500 studies
train_500 = train.head(500)

print("Total train rows:", len(train))
print("Selected rows:", len(train_500))
print("Train series rows:", len(train_series))
```

```python
print(os.listdir(DATA_DIR))
```

```python
print("Train columns:")
print(train_500.columns.tolist())

print("\nTrain Series columns:")
print(train_series.columns.tolist())
```

```python
print("Missing values in train_500:")
print(train_500.isnull().sum())

print("\nMissing values in train_series:")
print(train_series.isnull().sum())
```

```python
print("Duplicate StudyInstanceUID:",
      train_500["StudyInstanceUID"].duplicated().sum())

print("Duplicate SeriesInstanceUID:",
      train_series["SeriesInstanceUID"].duplicated().sum())
```

```python
print("Number of selected studies:", train_500["StudyInstanceUID"].nunique())

print("Number of series:", 
      train_series[
          train_series["StudyInstanceUID"].isin(
              train_500["StudyInstanceUID"]
          )
      ]["SeriesInstanceUID"].nunique())
```
